```python
# Create a class to control a smart light that can be turned ON or OFF and display its status.

class SmartLight:
    def __init__(self, name):
        self.name = name
        self.status = "OFF"

    def switch(self, action):
        self.status = action.upper()
        print(f"{self.name} is {self.status}")

light = SmartLight("Bedroom Light")
light.switch("ON")
```

```python
# Create a class that stores employee name, ID, and department and displays ID card details.

class Employee:
    def __init__(self, name, emp_id, dept):
        self.name = name
        self.emp_id = emp_id
        self.dept = dept

    def display(self):
        print("Employee ID Card")
        print(f"Name: {self.name}")
        print(f"ID: {self.emp_id}")
        print(f"Department: {self.dept}")

emp = Employee("Rahul", "EMP102", "IT")
emp.display()
```

```python
# Create a class to store contact name and phone number and display contact information.

class Contact:
    def __init__(self, name, phone):
        self.name = name
        self.phone = phone

    def save_contact(self):
        print("Contact Saved")
        print(f"Name: {self.name}")
        print(f"Phone: {self.phone}")

c1 = Contact("Anita", "9876543210")
c1.save_contact()
```

```python
# Create a class that stores product name and price and prints a formatted price tag.

class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def price_tag(self):
        print(f"Product: {self.name}")
        print(f"Price: ₹{self.price}")

p1 = Product("Headphones", 2499)
p1.price_tag()
```

```python
# Create a class that stores movie name and rating and displays movie details.

class Movie:
    def __init__(self, name, rating):
        self.name = name
        self.rating = rating

    def display(self):
        print(f"Movie: {self.name}")
        print(f"Rating: {self.rating} / 5")

m1 = Movie("Inception", 4.8)
m1.display()
```

```python
# Create a class that stores customer name and delivery address and prints delivery details.

class Delivery:
    def __init__(self, customer, address):
        self.customer = customer
        self.address = address

    def show_details(self):
        print("Delivery Details")
        print(f"Customer: {self.customer}")
        print(f"Address: {self.address}")

d1 = Delivery("Suman", "Hyderabad")
d1.show_details()
```
