```python
#Smart Parking Lot Management System
def smart_parking(capacity, logs):
    parked = 0
    peak = 0

    for log in logs:
        if log == "IN":
            parked += 1
        elif log == "OUT" and parked > 0:
            parked -= 1

        peak = max(peak, parked)

    status = "Full" if parked > capacity else "Available"

    print("Currently Parked Vehicles:", parked)
    print("Parking Status:", status)
    print("Peak Usage:", peak)
```

```python
#Online Food Delivery Time Estimator
def estimate_delivery_time(distance, traffic, weather):
    base_time = distance * 5   # 5 mins per km

    if traffic == "High":
        base_time += 20
    elif traffic == "Medium":
        base_time += 10

    if weather == "Rainy":
        base_time += 15
    elif weather == "Storm":
        base_time += 25

    print("Estimated Delivery Time:", base_time, "minutes")
```

```python
#Movie Theatre Seat Occupancy Analyzer
def theatre_occupancy(total_seats, booked_seats):
    booked = len(booked_seats)
    occupancy = (booked / total_seats) * 100

    if occupancy == 100:
        status = "Housefull"
    elif occupancy >= 75:
        status = "Almost Full"
    else:
        status = "Available"

    print(f"Occupancy: {occupancy:.0f}%")
    print("Show Status:", status)
```

```python
#Cloud Server Load Classification System
def server_load(cpu_readings):
    avg = sum(cpu_readings) / len(cpu_readings)

    if avg < 50:
        status = "Normal"
    elif avg <= 80:
        status = "Warning"
    else:
        status = "Critical"

    print(f"Average CPU Load: {avg:.0f}%")
    print("Server Status:", status)
```

```python
#Smart Classroom Resource Usage Monitor
def resource_monitor(usage):
    overused = []

    for resource, hours in usage.items():
        if hours > 8:
            overused.append(resource)

    print("Overused Resources:", ", ".join(overused) if overused else "None")
    print("Energy Alert:", "Yes" if overused else "No")
```

```python
#Online Event Registration Capacity Controller
def event_registration(capacity, registrations):
    confirmed = min(capacity, registrations)
    waitlisted = max(0, registrations - capacity)
    status = "Closed" if registrations >= capacity else "Open"

    print("Confirmed Registrations:", confirmed)
    print("Waitlisted Users:", waitlisted)
    print("Registration Status:", status)
```
