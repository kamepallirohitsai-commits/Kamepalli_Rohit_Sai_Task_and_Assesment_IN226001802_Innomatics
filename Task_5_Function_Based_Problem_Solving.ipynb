```python
# Create a function to analyze engagement on a social media post.
def post_engagement_analyzer(likes):
    total_likes = 0
    for like in likes:
        total_likes += like

    if total_likes >= 1000:
        status = "Viral Post"
    else:
        status = "Normal Engagement"

    print(f"Total Likes: {total_likes}")
    print(f"Post Status: {status}")


post_engagement_analyzer([250, 300, 200, 300])
```

```python
# Create a function to monitor medicine stock levels.
def medicine_stock_alert(stock):
    if stock < 10:
        status = "Low Stock Alert"
    else:
        status = "Stock Sufficient"

    print(f"Medicine Stock: {stock}")
    print(f"Status: {status}")


medicine_stock_alert(6)
```

```python
# Create a function that analyzes seasonal rainfall data.
def rainfall_checker(rainfall_data, required_level):
    total = 0
    for rain in rainfall_data:
        total += rain

    average_rainfall = total / len(rainfall_data)

    if average_rainfall >= required_level:
        status = "Adequate Rainfall"
    else:
        status = "Inadequate Rainfall"

    print(f"Average Rainfall: {average_rainfall}")
    print(f"Rainfall Status: {status}")


rainfall_checker([70, 80, 65, 75], 70)
```

```python
# Create a function to detect duplicate usernames in a system.
def duplicate_account_detector(usernames):
    unique_usernames = set(usernames)

    if len(usernames) != len(unique_usernames):
        status = "Yes"
    else:
        status = "No"

    print(f"Duplicate Accounts Found: {status}")


duplicate_account_detector(["rohit", "sai", "rohit"])
```

```python
# Create a function to check whether a patient is eligible for an appointment slot.
def appointment_eligibility(age):
    if age >= 18:
        status = "Eligible"
    else:
        status = "Not Eligible"

    print(f"Patient Age: {age}")
    print(f"Eligibility Status: {status}")


appointment_eligibility(21)
```

```python
# Create a function that filters crop prices above a premium threshold.
def premium_crop_filter(prices):
    premium_crops = []

    for price in prices:
        if price > 2000:
            premium_crops.append(price)

    print(f"Premium Crops: {premium_crops}")


premium_crop_filter([1500, 2500, 3200, 1800])
```

```python
# Create a function to evaluate application health based on error count.
def application_health_checker(errors):
    if errors == 0:
        status = "Healthy"
    elif errors <= 5:
        status = "Minor Issues"
    else:
        status = "Critical Issues"

    print(f"Error Count: {errors}")
    print(f"System Status: {status}")


application_health_checker(7)
```

```python
# Create a function to validate daily transaction amount.
def transaction_limit_checker(amount):
    if amount <= 50000:
        status = "Approved"
    else:
        status = "Rejected"

    print(f"Transaction Amount: {amount}")
    print(f"Transaction Status: {status}")


transaction_limit_checker(60000)
```

```python
# Create a function to calculate attendance percentage and check exam eligibility.
def attendance_eligibility(attended_classes, total_classes):
    percentage = (attended_classes / total_classes) * 100

    if percentage >= 75:
        status = "Eligible"
    else:
        status = "Not Eligible"

    print(f"Attendance Percentage: {percentage}")
    print(f"Exam Eligibility: {status}")


attendance_eligibility(40, 50)
```

```python
# Create a function that calculates the electricity bill and classifies usage level based on units consumed.
def electricity_bill_analyzer(units):
    if units <= 100:
        bill = units * 3
    elif units <= 200:
        bill = (100 * 3) + (units - 100) * 5
    else:
        bill = (100 * 3) + (100 * 5) + (units - 200) * 7

    if bill < 500:
        usage = "Low Usage"
    elif bill <= 1500:
        usage = "Moderate Usage"
    else:
        usage = "High Usage"

    print(f"Total Bill: {bill}")
    print(f"Usage Status: {usage}")


electricity_bill_analyzer(250)
```
